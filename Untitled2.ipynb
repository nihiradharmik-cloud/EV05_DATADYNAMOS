```python
import pandas as pd
data = pd.read_csv("/content/RA2019_A52 (3).csv", encoding='latin1')
data.head()
```

```python
data.columns
```

```python
data.info()
data.isnull().sum()
```

```python
num_cols = data.select_dtypes(include=['int64','float64']).columns
data[num_cols] = data[num_cols].fillna(data[num_cols].mean())

cat_cols = data.select_dtypes(include=['object']).columns
data[cat_cols] = data[cat_cols].fillna("Unknown")
```

```python
import matplotlib.pyplot as plt

data.iloc[:, 2].value_counts().plot(kind="bar")
plt.title("Sample Accident Data Distribution")
plt.show()
```
